```python
# Text tokenization & vocabulary creation & id and tokens conversion
# By default characteristics: lowercase, no punctuation, no accents, no emojis, with stop words
# ['user', 'url', 'hashtag', 'rt', 'amp', 'gt']

def tokenize_user(raw_text, tokenizer, language='english', custom_stop_words=[],
                  lowercase=True, stp_wrds=True, emojis=False, accents=False, punct=False):
  L = list()

  if stp_wrds:
    stop_words = set(stopwords.words(language))
    stop_words.update([word.translate(str.maketrans('', '', string.punctuation)) for word in stop_words])
    stop_words.update(custom_stop_words)

  if lowercase and (not emojis) and (not accents) and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(unidecode(letter).lower().translate(str.maketrans('', '', string.punctuation)), repl='')
        if stp_wrds:
          if aux_s not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and (not emojis) and (not accents) and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(unidecode(letter).lower(), repl='')
        if stp_wrds:
          if aux_s not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and (not emojis) and accents and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(letter.lower().translate(str.maketrans('', '', string.punctuation)), repl='')
        if stp_wrds:
          if unidecode(aux_s) not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and (not emojis) and accents and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(letter.lower(), repl='')
        if stp_wrds:
          if unidecode(aux_s) not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and emojis and (not accents) and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += unidecode(letter).lower().translate(str.maketrans('', '', string.punctuation))
        if stp_wrds:
          if aux_s not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and emojis and (not accents) and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += unidecode(letter).lower()
        if stp_wrds:
          if aux_s not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and emojis and accents and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += letter.lower().translate(str.maketrans('', '', string.punctuation))
        if stp_wrds:
          if unidecode(aux_s) not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif lowercase and emojis and accents and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += letter.lower()
        if stp_wrds:
          if unidecode(aux_s) not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and (not emojis) and (not accents) and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(unidecode(letter).translate(str.maketrans('', '', string.punctuation)), repl='')
        if stp_wrds:
          if aux_s.lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and (not emojis) and (not accents) and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(unidecode(letter), repl='')
        if stp_wrds:
          if aux_s.lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and (not emojis) and accents and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(letter.translate(str.maketrans('', '', string.punctuation)), repl='')
        if stp_wrds:
          if unidecode(aux_s).lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and (not emojis) and accents and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += demoji.replace(letter, repl='')
        if stp_wrds:
          if unidecode(aux_s).lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and emojis and (not accents) and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += unidecode(letter).translate(str.maketrans('', '', string.punctuation))
        if stp_wrds:
          if aux_s.lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and emojis and (not accents) and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += unidecode(letter)
        if stp_wrds:
          if unidecode(aux_s).lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and emojis and accents and (not punct):
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        aux_s = ''
        for letter in word:
          aux_s += letter.translate(str.maketrans('', '', string.punctuation))
        if stp_wrds:
          if unidecode(aux_s).lower() not in stop_words:
            L_aux.append(aux_s)
        else:
          L_aux.append(aux_s)
      L.append(L_aux)
    return L

  elif (not lowercase) and emojis and accents and punct:
    for sentence in raw_text:
      L_aux = list()
      for word in tokenizer.tokenize(sentence):
        if stp_wrds:
          if unidecode(word).lower() not in stop_words:
            L_aux.append(word)
        else:
          L_aux.append(word)
      L.append(L_aux)
    return L


def make_vocabulary(raw_text, tokenizer, language='english', custom_stop_words=[],
                    lowercase=True, stp_wrds=True, emojis=False, accents=False, punct=False):
  freq_vocab = dict()

  for user in raw_text:
    L = tokenize_user(user, tokenizer, language, custom_stop_words,
                      lowercase, stp_wrds, emojis, accents, punct)

    for sentence in L:
      for token in sentence:
        if token in freq_vocab:
          freq_vocab[token] +=1
        else:
          freq_vocab[token] = 1

  L = sorted(freq_vocab.items(), key=lambda x: x[1], reverse=True)

  vocab = dict((tok[0], id) for id, tok in enumerate(L))

  return vocab, freq_vocab

def load_vocabulary(vocab_file):
  d = dict()
  with open(vocab_file, "r", encoding="utf-8") as reader:
    tokens = reader.readlines()
  for id, token in enumerate(tokens):
    token = token.rstrip("\n")
    d[token] = id
  return d

def load_freq_vocabulary(freq_vocab_file):
  d = dict()
  with open(freq_vocab_file, "r", encoding="utf-8") as reader:
    tokens = reader.readlines()
  for token in tokens:
    token = token.split()
    d[token[0]] = token[1]
  return d

def tokens_of_ids(vocab):
  d = dict((id, tok) for tok, id in vocab.items())
  return d
```

```python
# Function for vectorizing the tokenized text of a user

def vectorize_user(tok_text, vocabulary, length=np.inf):
  if length == np.inf:
    length = len(vocabulary)
  v = [0] * length
  for sentence in tok_text:
    for token in sentence:
      if token in vocabulary:
        if vocabulary[token] < length:
          v[vocabulary[token]] = 1  # We access the id of the corresponding token in the vocabulary
  return v
```

```python
# Create one-hot encoding (300 and 768)

pd.options.mode.copy_on_write = True

new_stop_words = ['', 'user', 'url', 'hashtag', 'rt', 'amp', 'gt', 'lt']
vocab_english, freq_vocab_english = make_vocabulary(
    training_set_english['raw_text'], tokenizer=TreebankWordTokenizer(),
    language='english',custom_stop_words=new_stop_words
    )

sorted_vocab_english = sorted(freq_vocab_english.items(), key=lambda x: x[1], reverse=True)

with open(os.path.join(DATA_PATH, "vocab_english.txt"), 'w') as f:
  for element in sorted_vocab_english:
    f.write(element[0] + '\n')

with open(os.path.join(DATA_PATH, "freq_vocab_english.txt"), 'w') as f:
  for element in sorted_vocab_english:
    f.write(element[0] + ' ' + str(element[1]) + '\n')

df_sorted_vocab_english = pd.DataFrame(
    (sorted_vocab_english[i][1] for i in range(len(sorted_vocab_english)))
    ).T.set_axis(
        (sorted_vocab_english[i][0] for i in range(len(sorted_vocab_english))), axis=1
        )

df_sorted_vocab_english.to_csv(os.path.join(DATA_PATH, "sorted_vocab_english.csv"), index=False)

df_onehot300_english = df_sorted_vocab_english.iloc[:, :300]
df_onehot768_english = df_sorted_vocab_english.iloc[:, :768]

for user in training_set_english['raw_text']:
  v_300 = vectorize_user(tokenize_user(user, tokenizer=TreebankWordTokenizer(), language='english',
                                   custom_stop_words=new_stop_words),
                         vocabulary=vocab_english, length=300)
  v_768 = vectorize_user(tokenize_user(user, tokenizer=TreebankWordTokenizer(), language='english',
                                   custom_stop_words=new_stop_words),
                         vocabulary=vocab_english, length=768)
  df_onehot300_english.loc[len(df_onehot300_english)] = v_300
  df_onehot768_english.loc[len(df_onehot768_english)] = v_768
df_onehot300_english.insert(0, 'user_id', np.concatenate((np.array([None]), training_set_english['user'])))
df_onehot300_english.insert(1, 'user_language', np.concatenate((np.array([None]), training_set_english['language'])))
df_onehot300_english.insert(2, 'user_class', np.concatenate((np.array([None]), training_set_english['class'])))
df_onehot768_english.insert(0, 'user_id', np.concatenate((np.array([None]), training_set_english['user'])))
df_onehot768_english.insert(1, 'user_language', np.concatenate((np.array([None]), training_set_english['language'])))
df_onehot768_english.insert(2, 'user_class', np.concatenate((np.array([None]), training_set_english['class'])))

df_onehot300_english.to_csv(os.path.join(DATA_PATH, "one-hot300_english.csv"), index=False)
df_onehot768_english.to_csv(os.path.join(DATA_PATH, "one-hot768_english.csv"), index=False)
```
